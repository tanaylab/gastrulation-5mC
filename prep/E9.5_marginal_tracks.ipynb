# E9.5 marginal ATAC tracks

Builds genome-wide marginal ATAC tracks for the E9.5 DKO and host fractions. The karyotype check on them (ED 10h) is in `E9.5_multiome.Fig6/Fig6.ipynb`.
- Input: single-cell ATAC counts in `tg_atac_dir/mm_embryo/dnmt_dko_brain{,_host}/scc_counts`.
- Outputs: misha tracks `embdnmt.brain_e95_dko.marginal`, `embdnmt.brain_e95_host.marginal` (shared database, inputs of peak calling in `prep/E9.5_atac_peaks.ipynb`).

## Marginal tracks

`scc_to_marginal_track` sums all cells of each fraction into one track (plus `_normed` and `_smoothed` companions).
- Caveat: writes to the shared database with no guard; re-running fails while the tracks exist. The host cell was not run in the stored notebook.

```r
scc_dko = scc_read(paste0(tg_atac_dir, "/mm_embryo/dnmt_dko_brain/scc_counts/"))
scc_host = scc_read(paste0(tg_atac_dir, "/mm_embryo/dnmt_dko_brain_host/scc_counts/"))
```

```r
scc_to_marginal_track(
  sc_counts = scc_dko,
  track = "embdnmt.brain_e95_dko.marginal")
```

```r
scc_to_marginal_track(
  sc_counts = scc_host,
  track = "embdnmt.brain_e95_host.marginal")
```
