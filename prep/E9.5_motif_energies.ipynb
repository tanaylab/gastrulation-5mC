# Motif energies of the E9.5 CREs

Computes normalized PREGO motif energies for the E9.5 CREs with `iceqream::compute_motif_energies` (as in the Methods for gastrulation peaks, but on 500 bp instead of 300 bp intervals).

- Input: `./data/output/cre_peaks_for_energ_passed_threshold.RDS` (24,124 CREs of 500 bp, filtered from the `init_e95` CRE set in `E9.5_multiome.Fig6/Fig6_func.r`).
- Outputs: `out_path("./data/output/energ_cre_blm.RDS")`, `out_path("./data/output/energ_cre_fulldb.RDS")` (read by `init_e95`, `E9.5_multiome.Fig6/Fig6.ipynb`).
- The filtered input file was made by a cell of the original `revision_e95/Fig6.ipynb`, later commented out: CREs with accessibility max - min > 3 across E9.5 states. That code is not in this repo.

## Inputs

CREs: `data/output/cre_peaks_for_energ_passed_threshold.RDS`.

```r
cre_peaks_for_energ = readRDS("./data/output/cre_peaks_for_energ_passed_threshold.RDS")
```

## Motif energies

- BLM database (`motif_reg_dir/output/BLM_db.rds`, `blm$pssm_db`), 32 threads -> `energ_cre_blm.RDS`
- Default database -> `energ_cre_fulldb.RDS`
- An unrun cell saves the same full-database energies again as `energ_thresh3_fulldb.RDS`; nothing else reads it.

```r
blm <- readr::read_rds(paste0(motif_reg_dir, "/output/BLM_db.rds"))
```

```r
prego::set_parallel(32)
energ = iceqream::compute_motif_energies(peak_intervals = cre_peaks_for_energ, db = blm$pssm_db, normalize = T)
saveRDS(energ, out_path("./data/output/energ_cre_blm.RDS"))
```

```r
energ = iceqream::compute_motif_energies(peak_intervals = cre_peaks_for_energ, normalize = T)
saveRDS(energ, out_path("./data/output/energ_cre_fulldb.RDS"))
```

```r
saveRDS(energ, out_path("./data/output/energ_thresh3_fulldb.RDS"))
```
